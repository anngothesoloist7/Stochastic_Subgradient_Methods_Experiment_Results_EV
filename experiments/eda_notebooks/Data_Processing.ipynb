# 2. Processing charging stations in HCM

```python
# Charging stations in HCM
df_hcm_new = df_merged[df_merged["ma_tram"].str.contains("C.HCM")]

warnings.simplefilter(action='ignore', category=pd.errors.SettingWithCopyWarning)

# Parse datetime
df_hcm_new["thoi_gian_bat_dau"] = pd.to_datetime(df_hcm_new["thoi_gian_bat_dau"], format="mixed", dayfirst=True)
df_hcm_new["thoi_gian_dung_sac"] = pd.to_datetime(df_hcm_new["thoi_gian_dung_sac"], format="mixed", dayfirst=True)
df_hcm_new["thoi_gian_sac"] = df_hcm_new["thoi_gian_dung_sac"] - df_hcm_new["thoi_gian_bat_dau"]

df_hcm_new["thoi_gian_sac"] = pd.to_timedelta(df_hcm_new["thoi_gian_sac"])
df_hcm_new["thoi_gian_sac_hours"] = df_hcm_new["thoi_gian_sac"].dt.total_seconds() / 3600
```

## Number of porters per station and point

```python
# Number of porters per station and point
hcm_station = df_hcm_new.groupby(["ma_tram", "ma_tru"])["ma_cong"].nunique().reset_index()
hcm_station = hcm_station.rename(columns={"ma_cong": "so_cong"})

# Add number of points into df
temp = df_hcm_new.groupby("ma_tram")["ma_tru"].nunique().to_frame().reset_index()
hcm_station = pd.merge(hcm_station, temp, on="ma_tram", how="left")
hcm_station= hcm_station.rename(columns={"ma_cong": "so_tram", "ma_tru_y": "so_tru", "ma_tru_x": "ma_tru"})

hcm_station
```

```python
warnings.simplefilter(action='ignore', category=pd.errors.SettingWithCopyWarning)

df10to19_clean = df_hcm_new[(df_hcm_new["dien_nang_tieu_thu"] > 0) & (df_hcm_new["loai_giao_dich"] == "Tinh_phi")]

# Full of battery
df10to19_clean_daypin = df10to19_clean[df10to19_clean["li_do_dung_sac"] == "Day_Pin"]
# Compute power capacity
df10to19_clean_daypin["cong_suat_calculated"] = df10to19_clean_daypin["dien_nang_tieu_thu"] / df10to19_clean_daypin["thoi_gian_sac_hours"]
# Amount of charged battery
df10to19_clean_daypin["muc_pin_da_sac"] = df10to19_clean_daypin["muc_pin_ket_thuc"] - df10to19_clean_daypin["muc_pin_bat_dau"]

df10to19_clean_daypin["cong_suat_calculated"]
```

```python
columns_to_keep = [
    "ma_tram", "ma_tru", "ma_cong", "thoi_gian_sac_hours", "cong_suat_calculated", "cong_suat", "muc_pin_da_sac",
    "dien_nang_tieu_thu", "ma_giao_dich_tren_emsp", "thoi_gian_bat_dau", "thoi_gian_dung_sac",
    "thoi_gian_ket_thuc", "muc_pin_bat_dau", "muc_pin_ket_thuc", "so_khung", "dong", "dien_ap", "id_tag"]

# Remove unnecessary columns
df10to19_clean_daypin = df10to19_clean_daypin[columns_to_keep]
# Print out cleaned column
cols = df10to19_clean_daypin.columns.tolist()
list(enumerate(cols))
```

```python
tru_sac = pd.read_csv("/content/drive/MyDrive/EV_charging_project/datasets/tru_sac.csv")
power_dict = {
    "AC11": 11, "DC60_JUPT": 60, "CC_60kW_Mini": 60, "CC_22KW": 22, "AC22_AURORA": 22,
    "DC30": 30, "DC180_TTAN": 180, "DC60": 60, "CC_80kW_Mini": 80, "SC_60kW_V3": 60,
    "CC_120kW_Mini": 120, "CC_07kW": 7, "DC250": 250, "SC_150kW_V3": 150, "SC_120kW": 120,
    "SC_30kW_Venus": 30, "SC_300kW_NC_Nova": 300, "CC_60kW": 60, "SC_300kW_LC_Nova": 300,
    "DH_60kW": 60, "CC_150kW": 150, "SC_360kW_Neptune": 360, "RB_60kW": 60,
    "RB_150KW": 150, "RB_300KW": 300, "CC_30kW_Kern": 30, "AT_120kW": 120, "SC_80kW": 80
}
# Merge charger type by charger id to df10to19_clean_daypin
full_hcm_stations = df10to19_clean_daypin.merge(tru_sac[['charger_id', 'charger_type']],
                            left_on='ma_tru', right_on='charger_id', how='left')
# Map charger types to power capacities
full_hcm_stations['charger_power_kW'] = full_hcm_stations['charger_type'].map(power_dict)

full_hcm_stations
```

# 3. Charging Station at C.HCM0038

C.HCM0038	has the highest number of charging sessions count.
* Start date: 2024-12-31
* End date: 2025-01-31

```python
full_hcm_stations.groupby("ma_tram")["ma_tru"].count().sort_values(ascending=False)
```

```python
best_HCM_station = full_hcm_stations[full_hcm_stations["ma_tram"]=="C.HCM0038"]
```

```python
ev_mapping = {
    "1": "VF8", "2": "VF9", "3": "VF7", "9": "VF7", "4": "VFe34",
    "5": "VF5", "8": "VF6", "A": "VF6", "B": "VF3", "7": "Vinbus",
    "C": "Green Bus", "T": "MBC"
}

battery_capacity = {
    "VF3": 18.64, "VF5": 37.23, "VF6": 59.6, "VFe34": 42.0,
    "VF7": 59.6, "VF8": 87.7, "VF9": 123.0, "Vinbus": None,
    "Green Bus": None, "MBC": None
}

# Based on frame number
best_HCM_station ["EV_model"] = best_HCM_station ["so_khung"].str[4].map(ev_mapping)

# Mapping EV model to battery capacity
best_HCM_station ["EV_battery_capacity_kWh"] = best_HCM_station ["EV_model"].map(battery_capacity)

# Moving order of columns
columns_to_move = ["charger_type", "charger_power_kW", "EV_model", "EV_battery_capacity_kWh"]
new_column_order = columns_to_move + [col for col in best_HCM_station .columns if col not in columns_to_move]
best_HCM_station = best_HCM_station[new_column_order]

# Moving charging efficiency to the front
best_HCM_station["charging_efficiency"] = best_HCM_station["EV_battery_capacity_kWh"]*best_HCM_station["muc_pin_da_sac"] / (100*best_HCM_station["dien_nang_tieu_thu"])
best_HCM_station_ = best_HCM_station[["charging_efficiency"] + [col for col in best_HCM_station.columns if col != "charging_efficiency"]]
best_HCM_station.head()
```

```python
columns_to_keep = [
    "EV_battery_capacity_kWh", "charging_efficiency", "charger_type", "charger_power_kW", "EV_model", "ma_tram", "ma_tru", "ma_cong",
    "ma_giao_dich_tren_emsp", "thoi_gian_bat_dau", "thoi_gian_dung_sac", "thoi_gian_ket_thuc", "dien_nang_tieu_thu", "muc_pin_bat_dau",
    "muc_pin_ket_thuc", "dong", "dien_ap", "cong_suat", "id_tag", "so_khung",  "thoi_gian_sac_hours", "cong_suat_calculated", "muc_pin_da_sac"]

best_HCM_station_ = best_HCM_station[columns_to_keep]
```

```python
best_HCM_station_["thoi_gian_dung_sac"].sort_values()
```

```python
best_HCM_station_.info()
```

```python
best_HCM_station_["thoi_gian_bat_dau"] = pd.to_datetime(best_HCM_station_["thoi_gian_bat_dau"], format="mixed", dayfirst=True)
best_HCM_station_["thoi_gian_ket_thuc"] = pd.to_datetime(best_HCM_station_["thoi_gian_ket_thuc"], format="mixed", dayfirst=True)
best_HCM_station_["thoi_gian_dung_sac"] = pd.to_datetime(best_HCM_station_["thoi_gian_dung_sac"], format="mixed", dayfirst=True)
best_HCM_station_["thoi_gian_sac"] = best_HCM_station_["thoi_gian_dung_sac"] - best_HCM_station_["thoi_gian_bat_dau"]

best_HCM_station_["thoi_gian_sac"] = pd.to_timedelta(best_HCM_station_["thoi_gian_sac"])
best_HCM_station_["thoi_gian_sac_hours"] = best_HCM_station_["thoi_gian_sac"].dt.total_seconds() / 3600

best_HCM_station_["over_time"] = best_HCM_station_["thoi_gian_ket_thuc"] - best_HCM_station_["thoi_gian_dung_sac"]

# best_HCM_station_["over_time"] = pd.to_timedelta(best_HCM_station_["over_time"])
best_HCM_station_["over_time"] = best_HCM_station_["over_time"].dt.total_seconds() / 60
best_HCM_station_["over_time"]
```

```python
# Move over_time to the front
best_HCM_station_ = best_HCM_station_[["over_time"] + [col for col in best_HCM_station_.columns if col != "over_time"]]
# energy = capacity * battery level
best_HCM_station_["energy_charged"] = best_HCM_station_["EV_battery_capacity_kWh"] * best_HCM_station_["muc_pin_da_sac"] / 100

best_HCM_station_
```

```python
columns_to_keep = [
    "id_tag", "so_khung", "energy_charged", "EV_battery_capacity_kWh", "over_time", "charging_efficiency", "charger_type", "charger_power_kW",
    "EV_model", "ma_tram", "ma_tru", "ma_cong", "ma_giao_dich_tren_emsp", "dien_nang_tieu_thu", "muc_pin_bat_dau",
    "muc_pin_ket_thuc", "dong", "dien_ap", "cong_suat",  "thoi_gian_sac_hours", "cong_suat_calculated", "muc_pin_da_sac", 'thoi_gian_bat_dau', 'thoi_gian_dung_sac']
#'thoi_gian_bat_dau', 'thoi_gian_dung_sac'
best_HCM_station_ = best_HCM_station_[columns_to_keep]
best_HCM_station_ = best_HCM_station_[["energy_charged"] + [col for col in best_HCM_station_.columns if col != "energy_charged"]]
```
